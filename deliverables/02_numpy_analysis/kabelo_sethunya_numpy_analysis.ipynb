## 6. Relationship Between GDP Growth and Employment

The dataset contains GDP growth and employment observations for different countries, years, and economic sectors. NumPy can be used to calculate the numerical relationship between these two variables. The Pearson correlation coefficient is used to determine whether GDP growth and employment tend to move together in the dataset.

```python
# Calculate the Pearson correlation coefficient using NumPy

correlation_matrix = np.corrcoef(gdp_growth, employment)
correlation = correlation_matrix[0, 1]

print("GDP Growth vs Employment correlation:", correlation)
```

## 9. Standardised Numerical Analysis

Standardisation allows numerical observations to be compared relative to their mean and standard deviation. A z-score indicates how many standard deviations an observation is above or below the mean.

```python
# Calculate z-scores for GDP Growth and Employment

gdp_z_scores = (gdp_growth - np.mean(gdp_growth)) / np.std(gdp_growth)
employment_z_scores = (employment - np.mean(employment)) / np.std(employment)

print("GDP Growth z-scores (first 10):")
print(gdp_z_scores[:10])

print("\nEmployment z-scores (first 10):")
print(employment_z_scores[:10])
```

```python
# Identify observations more than two standard deviations from the GDP mean

extreme_gdp = np.abs(gdp_z_scores) > 2

print("GDP observations more than 2 standard deviations from the mean:",
      np.sum(extreme_gdp))

print("Percentage of extreme GDP observations:",
      np.mean(extreme_gdp) * 100, "%")
```
