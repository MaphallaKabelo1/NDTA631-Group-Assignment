```python
import pandas as pd
import numpy as np
```

```python
# Load the cleaned dataset
df = pd.read_csv("../../data/clean/GDP_Employment_Cleaned.csv")

df.head()
```

## 2. Numerical Data Preparation

The cleaned dataset contains 37,700 observations and five variables. For the NumPy numerical analysis, the numerical variables selected are `TIME_PERIOD`, `GDP_Growth`, and `Employment_Thousands`. The categorical variables `REF_AREA_LABEL` and `Sector` are used to identify countries and economic sectors but are not directly used for numerical calculations.

## 3. NumPy Array Conversion and Descriptive Statistics

```python
# Convert numerical columns to NumPy arrays

years = df["TIME_PERIOD"].to_numpy()
gdp_growth = df["GDP_Growth"].to_numpy()
employment = df["Employment_Thousands"].to_numpy()

print("Years array shape:", years.shape)
print("GDP Growth array shape:", gdp_growth.shape)
print("Employment array shape:", employment.shape)
```

```python
# Display the first 10 values from each NumPy array

print("Years:", years[:10])
print("GDP Growth:", gdp_growth[:10])
print("Employment:", employment[:10])
```

```python
# Descriptive statistics for GDP Growth

print("GDP Growth Mean:", np.mean(gdp_growth))
print("GDP Growth Median:", np.median(gdp_growth))
print("GDP Growth Minimum:", np.min(gdp_growth))
print("GDP Growth Maximum:", np.max(gdp_growth))
print("GDP Growth Standard Deviation:", np.std(gdp_growth))
print("GDP Growth Variance:", np.var(gdp_growth))
```

```python
# Descriptive statistics for Employment

print("Employment Mean:", np.mean(employment))
print("Employment Median:", np.median(employment))
print("Employment Minimum:", np.min(employment))
print("Employment Maximum:", np.max(employment))
print("Employment Standard Deviation:", np.std(employment))
print("Employment Variance:", np.var(employment))
```

### Interpretation

The NumPy descriptive statistics provide an overview of the distribution of GDP growth and employment values in the cleaned dataset. The mean and median provide measures of central tendency, while the minimum and maximum values show the range of the observations. Standard deviation and variance indicate the amount of variation in the numerical data.

These statistics provide a numerical foundation for understanding the relationship between economic growth and employment across countries, years, and economic sectors.

## 3. NumPy Array Conversion and Descriptive Statistics

The numerical variables were converted from Pandas Series into NumPy arrays to perform numerical analysis. NumPy was used to calculate measures of central tendency and dispersion, including the mean, median, minimum, maximum, standard deviation, and variance.

The GDP Growth results show a mean of approximately 4.05% and a median of approximately 3.95%. The standard deviation of approximately 4.43 indicates variation in GDP growth across the observations.

Employment shows a much larger difference between its mean and median. The mean employment value is approximately 5,222.36 thousand, while the median is approximately 339.67 thousand. This large difference, together with the maximum value of approximately 775,772.06 thousand, indicates substantial variation in employment values across the observations.

```python
# Calculate additional NumPy statistics

gdp_range = np.ptp(gdp_growth)
employment_range = np.ptp(employment)

gdp_percentiles = np.percentile(gdp_growth, [25, 50, 75])
employment_percentiles = np.percentile(employment, [25, 50, 75])

print("GDP Growth Range:", gdp_range)
print("GDP Growth 25th, 50th and 75th Percentiles:", gdp_percentiles)

print("\nEmployment Range:", employment_range)
print("Employment 25th, 50th and 75th Percentiles:", employment_percentiles)
```
